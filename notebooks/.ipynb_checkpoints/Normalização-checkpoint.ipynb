```python
import json
import pandas as pd
```

```python
autores_cd = json.load(open("../data/autores_proposicoes.json", "r", encoding="utf-8"))
autores_cd
```

```python
autores_dict = {}
for i in autores_cd[:1000]:
    autores_dict[int(i["id"])] = {"autores": i["autores"]}
```

```python
help(json.dump)
```

```python
with open("../data/camara/autores_proposicoes.json", 'w', encoding='utf-8') as f:
    json.dump(autores_dict, f, ensure_ascii=False, indent=2)
```
